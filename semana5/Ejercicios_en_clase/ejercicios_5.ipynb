```python
def suma_tupla(t):
    total = 0
    for e in t:
        total += e
    return total 

calorias = (800, 950, 700, 1300, 800, 900, 650)
resultado = suma_tupla(calorias)
print(f"metiste a tu cuerpo {resultado} calorias gordo")
```

```python
candidatos = {
    "A" : 0,
    "B" : 0,
    "C" : 0
}
cantidad = int(input("cuabtos votos van a ser?"))
conteo = 1

while conteo <= cantidad:
    voto = input(f"ingresa el voto {conteo} (A, B o c): ")
    if voto.upper() in candidatos:
        print("votos {conteo}")
        print(f"voto valido para {voto.upper()}")
        candidatos[voto.upper()] += 1
        conteo += 1
    else:
        print("el voto no es valido, vota nuevamente")

print("\n resultados")        
for k, v in candidatos.items():
    print(f"opcion {k}, votos: {v}")
    
```
