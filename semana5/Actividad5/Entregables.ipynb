## Agenda de contactos con busqueda

```python
agenda = {
    "Ana": "555-0101",
    "Luis": "555-0102",
    "Mía": "555-0103"
}


nombre = input("Nombre del nuevo contacto: ")
telefono = input("Teléfono del nuevo contacto: ")


agenda[nombre] = telefono

print(agenda.items())

def buscar_telefono(agenda, nombre):
    return agenda[nombre]

buscar = input("Nombre a buscar: ")


telefono_encontrado = buscar_telefono(agenda, buscar)

print("El teléfono de", buscar, "es:", telefono_encontrado)



```

## Contador de votos con diccionario

```python
votos = {"Rojo": 0, "Azul": 0, "Verde": 0}

cantidad = int(input("¿Cuántos votos quieres registrar? "))

for i in range(cantidad):
    color = input("¿A quién le das tu voto? (Rojo, Azul o Verde): ")
    votos[color] += 1

ganador = ""
mayor = 0

for color, total in votos.items():
    if total > mayor:
        ganador = color
        mayor = total

print(f"El color ganador es: {ganador}")
print(f"total de votos: {mayor}")

```

## Validacion robusta de entrada

```python
while True:
    try:
        numero1 = int(input("Ingresa el primer número entero: "))
        numero2 = int(input("Ingresa el segundo número entero: "))
        break
    except ValueError:
        print("Error: debes ingresar números enteros. Inténtalo de nuevo.")

while True:
    try:
        resultado = numero1 / numero2
        print("El resultado de la división es:", resultado)
        break
    except ZeroDivisionError:
        print("Error: no se puede dividir entre cero.")
        numero2 = int(input("Ingresa otro número diferente de cero: "))
```

## Analizador de textos

```python

def contar_palabras(frase):
    return len(frase.split())


frase = input("Ingresa una frase: ")

print(f"Longitud de la frase: {len(frase)}")

frase_mayusculas = frase.upper()
print("En mayúsculas:", frase_mayusculas)

frase_reemplazada = frase.replace("Python", "programación")
print("Frase reemplazada:", frase_reemplazada)

cantidad = contar_palabras(frase)
print("Cantidad de palabras:", cantidad)
```

```python

numeros = (8, 3, 10, 1, 5)

contactos = {
    "Juan": "5551234567",
    "Ana": "5559876543",
    "Pedro": "5554567890"
}


def mostrar_tupla():
    print("Tupla ordenada:")
    print(sorted(numeros))


def buscar_telefono():
    nombre = input("Ingresa el nombre del contacto: ")

    if nombre in contactos:
        print(f"Teléfono: {contactos[nombre]}")
    else:
        print("Contacto no encontrado.")



def dividir_numeros():
    try:
        numero1 = float(input("Ingresa el primer número: "))
        numero2 = float(input("Ingresa el segundo número: "))

        resultado = numero1 / numero2
        print("Resultado:", resultado)

    except ZeroDivisionError:
        print("Error: no se puede dividir entre cero.")
    except ValueError:
        print("Error: debes ingresar números válidos.")


def analizar_mensaje():
    mensaje = input("Ingresa un texto: ")

    print(f"Longitud del mensaje: {len(mensaje)}")
    print(f"Mensaje en mayúsculas: {mensaje.upper()}")
    print(f"Cantidad de palabras: {len(mensaje.split())}")


while True:
    print("= MENÚ PRINCIPAL =")
    print("1. Mostrar tupla ordenada")
    print("2. Buscar teléfono")
    print("3. Dividir dos números")
    print("4. Analizar mensaje")
    print("5. Salir")

    opcion = input("Selecciona una opción: ")

    if opcion == "1":
        mostrar_tupla()

    elif opcion == "2":
        buscar_telefono()

    elif opcion == "3":
        dividir_numeros()

    elif opcion == "4":
        analizar_mensaje()

    elif opcion == "5":
        print("Programa finalizado.")
        break

    else:
        print("Opción no válida. Intenta nuevamente.")

```
