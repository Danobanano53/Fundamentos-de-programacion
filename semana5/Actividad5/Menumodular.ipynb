## Menu modular

```python

def sumar_tupla(numeros):
    suma = sum(numeros)
    return suma



def buscar_contacto(contactos, nombre):
    if nombre in contactos:
        return contactos[nombre]
    else:
        return None



def contar_palabras(mensaje):
    palabras = mensaje.split()
    return len(palabras)



def seccion_tuplas():

    numeros = (10, 20, 30, 40, 50)

    print("TUPLAS")

   
    print(f"Tercer elemento: {numeros[2]}")

    
    numero1 = int(input("Ingresa un número adicional: "))
    numero2 = int(input("Ingresa otro número adicional: "))

    
    nueva_tupla = numeros + (numero1, numero2)

    print(f"Nueva tupla: {nueva_tupla}")

    lista = list(nueva_tupla)
    lista.sort()

    print(f"Lista ordenada: {lista}")

   
    resultado = sumar_tupla(nueva_tupla)

    print(f"Suma de todos los números: {resultado}")


def seccion_diccionarios():

    contactos = {
        "Juan": "4421234567",
        "Maria": "4427654321",
        "Carlos": "4429876543"
    }

    print("--- DICCIONARIOS ---")

    
    nombre_nuevo = input("Ingresa el nombre del nuevo contacto: ")
    telefono_nuevo = input("Ingresa su número de teléfono: ")

    contactos[nombre_nuevo] = telefono_nuevo

    print("Contactos registrados:")

    
    for nombre in contactos:
        print(nombre)


    nombre_buscar = input("¿Qué contacto quieres buscar?: ")

    telefono = buscar_contacto(contactos, nombre_buscar)

    if telefono != None:
        print(f"Teléfono: {telefono}")
    else:
        print("El contacto no existe.")



def seccion_excepciones():

    print("--- EXCEPCIONES ---")

    try:

        numero1 = int(input("Ingresa un número entero: "))
        numero2 = int(input("Ingresa otro número entero: "))

        suma = numero1 + numero2

        print(f"La suma es:{suma}")

        
        resultado = numero1 / numero2

        print(f"La división es: {resultado}")

    except ValueError:
        print("Error: debes ingresar números enteros")

    except ZeroDivisionError:
        print("Error: no se puede dividir entre cero")


def seccion_strings():

    print("--- STRINGS ---")

    mensaje = "Python es un lenguaje de programación"

   
    print(f"Mensaje: {mensaje}")
    print(f"Longitud del mensaje: {len(mensaje)}")

    mayusculas = mensaje.upper()

    print(f"En mayúsculas: {mayusculas}")

  
    nuevo_mensaje = mensaje.replace("Python", "Java")

    print(f"Mensaje reemplazado: {nuevo_mensaje}")


    cantidad = contar_palabras(mensaje)

    print(f"Cantidad de palabras: {cantidad}")


while True:
    print("====MENÚ PRINCIPAL=====")
    print("1. Tuplas")
    print("2. Diccionarios")
    print("3. Excepciones")
    print("4. Strings")
    print("5. Finalizar")

    opcion = input("Selecciona una opción: ")

    if opcion == "1":
        seccion_tuplas()

    elif opcion == "2":
        seccion_diccionarios()

    elif opcion == "3":
        seccion_excepciones()

    elif opcion == "4":
        seccion_strings()

    elif opcion == "5":
        print("Programa finalizado.")
        break

    else:
        print("Opción no válida.")
```
