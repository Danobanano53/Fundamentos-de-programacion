## suma con centinela 0

```python
suma = 0
while True:
    num = float(input("ingresa un valor(pon 0 para terminar): "))
    if num == 0:
        break
    suma += num
print(f"la suma de los valores es: {suma}")
```

##  decision para descuento por monto

```python
monto = float(input("INGRESE LA EL MONTO DE COMPRA: "))

if monto >= 300:
    descuento = monto * 0.10
    total = monto - descuento
    print(f"monto: {monto:.2f}")
    print(f"descuento: {descuento:.2f}")
    print(f"total a pagar: {total:.2f}")
else:
    print(f"total a pagar: {monto:.2f}")
```
