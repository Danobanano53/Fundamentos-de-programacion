```python
def promedio(sumatoria, n):
       return sumatoria / n

n = int(input("cuantas calificaciones: "))

sumatoria = 0

print(f"se van a pedir {n} calificaciones")

for c in range(1, n+1):
       calif = float(input(f"dame la calificacion {c}"))    
       sumatoria = sumatoria + calif
       print(f"hasta este punto la sumatoria va en {sumatoria}")

print(f"el promedio de {n} calificaciones es {promedio(sumatoria, n)}")
```

```python
filas = int(input("cuantas filas: "))
columnas = 3


matriz = []
for f in range(filas):
    fila_actual = []
    for c in range(columnas):
        #matriz[f][c] = int(input(f"valor para[{f}][{c}]"))
        valor = int(input(f"valor para[{{f}}][{{c}}]"))
        fila_actual.append(valor)
    matriz.append(fila_actual)

print("impresion de matriz")
for f in range(filas):
    for c in range(columnas):
        print(matriz[f][c], end=" ")
    print("")
```

```python
lista=[]
contador = 0
while contador <=10:
    valor = int(input("ingresa un numero entero entre -10 y 10: "))
    if valor >= -10 or valor <= 10:
     contador += 1
     lista.append(valor)

    else:
       print("valor incorrecto, ingresa un numero entre -10 y 10")

print(lista)
for i in lista:
   if i < 0:
        for e in range(-10):
            print(" ", end="")
        print("", end="") 
        for a in range(-i):
            print("", end="")
        print()
   elif i > 0:
        for e in range(10):
            print(" ", end="")
        print("|", end="")
        for a in range(i):
            print("*", end="")
        print()
   else: 
      print("|")
        
```
