## Suma de todos lo elemnos de una matriz

```python
tabla = [
    [1, 2, 3],
    [4, 5, 6],
    [7 ,8, 8]
]

def imprimir_tabla(tabla):
    for fila in tabla:
        for numero in fila:
            print(numero, end=" ")

    print()



def suma_tabla(tabla):
    suma = 0
    for fila in tabla:
        for numero in fila:
            suma += numero
    return suma   

imprimir_tabla(tabla)

resultado = suma_tabla(tabla)
print(f"la suma de los numeros en la matriz es: {resultado}")

```

## Funcion suma fila

```python
tabla = [
    [4, 6, 7],
    [8, 2, 5],
    [7, 8, 1]
]

def imprimir_tabla(tabla):
    for fila in tabla:
        for numero in fila:
            print(numero, end=" ")
        print()
def suma_fila1(tabla):
    suma = 0
    for numero in tabla[0]:
        suma += numero
    return suma
def suma_fila2(tabla):
    suma = 0
    for numero in tabla[1]:
        suma += numero
    return suma
def suma_fila3(tabla):
    suma = 0
    for numero in tabla[2]:
        suma += numero
    return suma

imprimir_tabla(tabla)
print(f"la suma de la fila 1 es: {suma_fila1(tabla)}")
print(f"la suma de las fila 2 es: {suma_fila2(tabla)}")
print(f"la suma de la fila 3 es: {suma_fila3(tabla)}")

        
    
```

## Tabla pitagoras tamaño variable

```python
rango = int(input("ingrese el numero de rango"))

tabla = []

for i in range(rango):

    fila = []
    for j in range(rango):
        resultado = i * j
        fila.append(resultado)

    tabla.append(fila)

def imprimir_tabla(tabla):

    for fila in tabla:
        for numero in fila:
            print(numero, end="\t")

        print()    

def buscar_coordenada(tabla,renglon,columna):

    resultado = tabla[renglon - 1][columna - 1]

    return resultado 

imprimir_tabla(tabla)

renglon = int(input(f"ingrese el primer factor (1 al {rango}):  "))
columna = int(input(f"ingrese el segundo factor (1 al {rango}):  "))
producto = buscar_coordenada(tabla, renglon, columna )

print(renglon, "x", columna, "=", producto)
```
