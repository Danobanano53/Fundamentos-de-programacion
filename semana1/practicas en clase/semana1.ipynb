## conversor de temperatura

```python
celsius = float(input("ingresa la temperatura en celsius: "))
fahrenheit = (celsius * 9/5) + 32
kelvin = celsius + 273.15
print("la temperatura en fahrenheit es:", fahrenheit)
print("la temperatura en kelvin es:", kelvin)
```

## CALIFICACION

```python
cal1 = float(input("ingresa la calificacion 1: "))
cal2 = float(input("ingresa la calificacion 2: "))
cal3 = float(input("ingresa la calificacion 3: "))
promedio = (cal1 + cal2 + cal3) / 3
print("el promedio es:", f"{promedio:.2f}")
print("me estrese mucho haciendo este programa, pero lo logre")
```

## convertir kilometros a millas 

```python
kilometros = float(input("ingresa la distancia en kilometros: "))
millas = kilometros * 0.621371
print("la distancia en millas es:", f"{millas:.2f}")
```
