## Division de cuenta con propina 

```python
cuenta = float(input("ingresa el total de la cuenta: "))
porcentaje_propina = float(input("ingresa el porcentaje de propina: "))
personas = int(input("ingresa el numero de personas: "))

propina = cuenta * (porcentaje_propina / 100)

total_con_propina = (cuenta + propina)
pago_por_persona = (total_con_propina / personas)

print(f"el monto de la propina es ${propina:.2f}")
print(f"el total con propina es ${total_con_propina:.2f}")
print(f"cada persona debe pagar ${pago_por_persona:.2f}")
```

## Programa para convertir minutos a días, horas y minutos

```python
minutos = int(input("ingresa la cantidad de minutos: "))

dias = minutos // 1440
restantes = minutos % 1440
horas = restantes // 60
minutos_restantes = restantes % 60

print(f"Días: {dias}")
print(f"Horas: {horas}")
print(f"Minutos: {minutos_restantes}")
```

## calificacion de 3 parciales

```python
cal1 = float(input("ingresa la calificacion 1: "))
cal2 = float(input("ingresa la calificacion 2: "))
cal3 = float(input("ingresa la calificacion 3: "))
promedio = (cal1 + cal2 + cal3) / 3
print("el promedio es:", f"{promedio:.2f}")
```

## Pesos a euro y dolar 

```python
pesos = float(input("ingresa la cantidad de pesos: "))
dolar = 17.02
euro = 19.71
print(f"Pesos mexicanos: ${pesos:.2f}")
print(f"Dólares (USD): ${round(pesos / dolar, 2)}")
print(f"Euros (EUR): €{round(pesos / euro, 2)}")
```

## Calculadora de tiempo en pantalla

```python
print("ingresa los datos que se piden a continuacion")
redes = float(input("1. Redes sociales: "))
videos = float(input("2. Videos / YouTube / TikTok: "))
juegos = float(input("3. Videojuegos: "))
mensajes = float(input("4. Mensajes / WhatsApp: "))
otras = float(input("5. Otras aplicaciones: "))

total_horas = redes + videos + juegos + mensajes + otras
porcentaje = (total_horas / 24) * 100

print(f"Tiempo total en el celular: {total_horas:.2f} horas")
print(f"Porcentaje de las 24 horas usado en el celular: {porcentaje:.2f}%")
```
