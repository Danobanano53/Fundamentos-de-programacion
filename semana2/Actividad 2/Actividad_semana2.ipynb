## Entrada al museo 

```python
cantidad = int(input("ingresa la cantidad de visitantes: "))

visitantes = []

i = 0


while i < cantidad:
   edad = int(input("ingrese la edad del  visitante: "))
   if edad < 0:
          print("edad no valida")
          continue
   elif edad <  3:
          precio = 0
   elif edad <= 17:
           precio = 30
   else:
          precio = 45

   tipo = input("ingresa el tipo de visitante(adulto_mayor, profesor, estudiante, otro): ")

   if tipo == "adulto_mayor":
    descuento = 0.12
   elif tipo == "profesor":
      descuento = 0.10
   elif tipo == "estudiante":
      descuento = 0.10
   else:
      descuento = 0

   precio_descuento = precio * descuento

   precio_final = precio - precio_descuento

   visitantes.append([edad, precio, tipo, descuento, precio_final ])

   i = i + 1
total = 0
for visitante in visitantes:
    print("Edad:", visitante[0])
    print("precio base:$", visitante[1])
    print("tipo de visitante:", visitante[2])
    print("descuento:$", visitante[3])
    print("precio final: $", visitante[4])
    
    total += visitante[4]
    
print("total final a pagar:$", total)

```
