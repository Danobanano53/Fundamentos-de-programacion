## Entrada al museo 

```python
cantidad = int(input("ingresa la cantidad de visitantes: "))

i = 0
total = 0

while i < cantidad:

    edad = int(input("ingrese la edad del visitante: "))

    if edad < 0:
        print("edad no valida")
        continue

    elif edad < 3:
        precio = 0

    elif edad <= 17:
        precio = 30

    else:
        precio = 45

    tipo = input("ingresa el tipo de visitante(adulto_mayor, profesor, estudiante, otro): ")

    if tipo == "adulto_mayor":
        descuento = 0.12

    elif tipo == "profesor":
        descuento = 0.10

    elif tipo == "estudiante":
        descuento = 0.10

    else:
        descuento = 0

    precio_descuento = precio * descuento
    precio_final = precio - precio_descuento

    total += precio_final

    i = i + 1

    print(f"visitante {i}: edad: {edad} tipo visitante: {tipo} descuento: %{descuento:.2f} descuento: ${precio_descuento:.2f} total final: ${precio_final:.2f}")

print("total final a pagar: $", total)
```

## Estadistica de visitantes con for

```python

cantidad = int(input("Ingresa la cantidad de visitantes: "))
adultos = 0
suma = 0

for i in range(cantidad):
    edad = int(input("ingresa la edad: "))

    print(f"visitante {i + 1}: {edad} años")
    if edad >= 18:
        adultos = adultos + 1


    suma += edad

    promedio = suma / cantidad
print(f"promedio de edad es: {promedio:.2f}")

print(f"cantidad de visitantes: {cantidad}")

print(f"la cantidad de adultos es: {adultos}")




 
     
```

## Control de aforo con break y continue

```python
total = 0
i = 1

while True:

    costo_boleto = float(input(f"costo del boleto{i}: "))

    if costo_boleto == 0:
        print("menor de 3 años, sin cargos")
        i = i + 1
        continue

    total = total + costo_boleto
    i = i + 1

    if total >= 100:
        print("aforo maximo alcanzado")
        break
print(f"Total final acumulado: ${total:.2f}")
```
