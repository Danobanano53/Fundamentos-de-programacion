## Descuelto por tipo de visitante

```python
tipo = input("Ingresa el tipo de visitante (adulto_mayor, profesor, estudiante u otro): ")

precio = 45

if tipo == "adulto_mayor":
    descuento = 0.12
elif tipo == "profesor":
    descuento = 0.10
elif tipo == "estudiante":
    descuento = 0.10
else:
    descuento = 0

precio_final = precio - (precio * descuento)

print("Precio final: $", round(precio_final, 2))
```
