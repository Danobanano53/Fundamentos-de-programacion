## Ejercicio 1

```python
print("===Area de un triagulo===")

base = float(input("ingresa la base del triangulo: "))
altura = float(input("ingresa la altura del triangulo: "))

print(f"la base es: {base}")
print(f"la altura es: {altura}")

area = (base * altura) / 2

print(f"El area del triagulo es {area}")
```

## Ejercicio 9

```python
def comparar(n1, n2, n3):
    if n1 >= n2 and n1 >= n3:
       return n1
    elif n2 >= n1 and n2 >= n3:
        return n2
    else:
        return n3

n1 = int(input("ingrese el primer valor"))
n2 = int(input("ingrese el segundo valor"))
n3 = int(input("ingrese el segundo valor"))

mayor = comparar(n1, n2, n3)

print(f"el mayor es {mayor}")

```

## Ejercicio 13

```python
def sumar_lista(lista):
    suma = 0

    for numero in lista:
        suma = suma + numero

    return suma


lista = [10, 20, 30, 40]

resultado = sumar_lista(lista)

print(f"La suma de los elementos es: {resultado}")
```

## Ejercicio 14

```python
def mayor_lista(lista):
    mayor = lista[0]

    for numero in lista:
        if numero > mayor:
            mayor = numero

    return mayor


lista = [7, 3, 9, 2]

resultado = mayor_lista(lista)

print(f"El elemento máximo es: {resultado}")
```

## Ejercicio 16

```python
def promedio(lista):
    suma = 0

    for numero in lista:
        suma = suma + numero

    promedio = suma / len(lista)

    print("El promedio es:", format(promedio, ".2f"))


lista = [8, 9, 10, 7]

promedio(lista)
```

## Ejercicio 18

```python
def factorial(n):
    if n == 0 or n == 1:
        return 1
    else:
        return n * factorial(n - 1)


numero = int(input("Ingresa un número: "))

resultado = factorial(numero)

print("El factorial de", numero, "es:", resultado)
```

## Ejercicio 21

```python
calificaciones = {
    "Ana": [8, 9, 10],
    "Beto": [7, 7, 8]
}

for nombre in calificaciones:
    notas = calificaciones[nombre]

    suma = 0

    for nota in notas:
        suma = suma + nota

    promedio = suma / len(notas)

    print(nombre + ":", format(promedio, ".2f"))
```
